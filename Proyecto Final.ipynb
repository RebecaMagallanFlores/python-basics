## Ordenando filas de un DataFrame
Ordene los datos de manera descendente de acuerdo a la columna Mujeres. ¿Cuáles son los 5 estados con mayor población de mujeres?

```python
#Ordene los datos
datos_ordenados=datos.sort_values(by='Mujeres', ascending=False)
display(datos_ordenados.head(5))
```

## Filtrando un DataFrame
Filtre el DataFrame para saber los Estados con población mayor a 5,000,000 de habitantes. ¿Cúal es el Estado con Mayor población?

```python
#Filtre los datos
datos_pob=datos[datos['Total']>5000000]
ordenados=(datos_pob.sort_values(by='Total', ascending=False))
display(ordenados)
display(f"Los estados con población mayor a 5 millones son: {list(ordenados['Estado'])}")
```

```python
#Los estados con población mayor a 5,000,000 de habitantes son: Ciudad de México, Jalisco, Nuevo León, Puebla, Veracruz
#El Estado con mayor población es: Ciudad de México
```

## Filtrando variables categóricas
Genere una lista llamada <code>Frontera</code> que contenga los Estados que tengan frontera con Estados Unidos:
Baja California, Sonora, Chihuahua, Coahuila, Nuevo León y Tamaulipas. Filtre el DataFrame con esta lista para conocer el Estado con mayor cantidad de Hombres que comparte frontera con Estados Unidos.


```python
#Genere una lista y filtre.
Frontera = ['Baja California', 'Sonora', 'Chihuahua', 'Coahuila', 'Nuevo León', 'Tamaulipas']
```

```python
# El Estado en la frontera con mayor cantidad de Hombres es:__________.
datos_frontera=datos[datos['Estado'].isin(Frontera)]
display(datos_frontera.sort_values(by='Hombres', ascending=False).head(1))
```

## Agregando nuevas columnas
Genere dos nuevas columnas llamadas <code>% Hombres</code> y <code>% Mujeres</code> que muestren la proporción de Hombres y Mujeres por Estado.

```python
#Genere dos nuevas columnas.
datos["% Hombres"] = datos["Hombres"] / datos["Total"] * 100
datos["% Mujeres"] =datos["Mujeres"] / datos["Total"] * 100
```
